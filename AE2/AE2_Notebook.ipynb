```python
import sys
import os

# Agregar la carpeta src al path para poder importar las clases
sys.path.append(os.path.abspath('src'))

from manejador_csv import ManejadorCSV
from manejador_json import ManejadorJSON
from generador_resumen_txt import GeneradorResumenTXT

# 1. Referencia las URLs específicas de origen (REQUERIDO POR LA RÚBRICA)
URL_ORIGEN_CSV = "https://huggingface.co/datasets/scikit-learn/iris/resolve/main/Iris.csv"
URL_ORIGEN_JSON = "https://huggingface.co/datasets/datasets-examples/doc-formats-json-1/resolve/main/data.json"

print("✅ Clases importadas y URLs configuradas correctamente.")
```

```python
# 2. Inicializar e importar CSV
csv_handler = ManejadorCSV('data/original.csv', 'data/modificado.csv')
encabezados, datos_originales = csv_handler.importar()

print("=== ANTES (CSV) ===")
print(f"Encabezados: {encabezados}")
print(f"Primera fila original: {datos_originales[0]}")
print(f"Total de registros: {len(datos_originales)}")

# 3. Modificar CSV (Cambiaremos la especie de la primera fila)
columna_a_modificar = "Species"
valor_anterior, valor_nuevo = csv_handler.modificar(columna_a_modificar, "Iris-setosa-MODIFICADA", fila=0)

print("\n=== DESPUÉS (CSV) ===")
print(f"Modificación: En columna '{columna_a_modificar}', fila 0, cambiamos '{valor_anterior}' por '{valor_nuevo}'")
print(f"Primera fila modificada: {csv_handler.datos[0]}")

# 4. Guardar nuevo CSV
csv_handler.guardar()
print("\n✅ Nuevo CSV guardado en: data/modificado.csv")
```

```python
# 2. Inicializar e importar JSON
json_handler = ManejadorJSON('data/original.json', 'data/modificado.json')
datos_json_originales = json_handler.importar()

print("=== ANTES (JSON) ===")
print(f"Primer registro original: {datos_json_originales[0]}")
total_registros_json = len(datos_json_originales) if isinstance(datos_json_originales, list) else 1
print(f"Total de registros: {total_registros_json}")

# 3. Modificar JSON (Agregar clave a todos los registros)
json_handler.modificar("procesado_por", "AE2_Evidencia_Aprendizaje")

print("\n=== DESPUÉS (JSON) ===")
print(f"Primer registro modificado: {json_handler.datos[0]}")

# 4. Guardar nuevo JSON
json_handler.guardar()
print("\n✅ Nuevo JSON guardado en: data/modificado.json")
```

```python
# 1. Inicializar la tercera clase con las URLs de origen
resumen_handler = GeneradorResumenTXT(URL_ORIGEN_CSV, URL_ORIGEN_JSON)

# 2. Registrar la operación del CSV
resumen_handler.registrar_operacion(
    formato="CSV",
    archivo_origen="data/original.csv",
    archivo_salida="data/modificado.csv",
    modificacion=f"Actualización de columna '{columna_a_modificar}' en la fila 0",
    registros=len(csv_handler.datos)
)

# 3. Registrar la operación del JSON
resumen_handler.registrar_operacion(
    formato="JSON",
    archivo_origen="data/original.json",
    archivo_salida="data/modificado.json",
    modificacion="Adición de la clave 'procesado_por' a todos los registros",
    registros=total_registros_json
)

# 4. Generar y mostrar el contenido del TXT
ruta_resumen = 'data/resumen.txt'
resumen_handler.generar_resumen(ruta_resumen)

print(f"✅ Resumen generado exitosamente en: {ruta_resumen}\n")
print("="*40)
print("CONTENIDO DEL ARCHIVO resumen.txt:")
print("="*40)
with open(ruta_resumen, 'r', encoding='utf-8') as f:
    print(f.read())
```
