```python
import sys
import os

# Agregar la carpeta src al path para poder importar las clases
sys.path.append(os.path.abspath('src'))

from manejador_csv import ManejadorCSV
from manejador_json import ManejadorJSON
from generador_resumen_txt import GeneradorResumenTXT

# 1. Referencia las URLs específicas de origen (REQUERIDO POR LA RÚBRICA)
URL_ORIGEN_CSV = "https://huggingface.co/datasets/scikit-learn/iris/resolve/main/Iris.csv"
URL_ORIGEN_JSON = "https://huggingface.co/datasets/datasets-examples/doc-formats-json-1/resolve/main/data.json"

print("✅ Clases importadas y URLs configuradas correctamente.")
```

```python
# Inicializar e importar CSV
csv_handler = ManejadorCSV('data/original.csv', 'data/modificado.csv')
encabezados, datos_originales = csv_handler.importar()
registros_antes_csv = len(csv_handler.datos)

print("=== ANTES (CSV) ===")
print(f"Encabezados: {encabezados}")
print(f"Primera fila: {csv_handler.datos[0]}")
print(f"Total de registros: {registros_antes_csv}")

# Modificación 1: actualizar un valor
columna = "Species"
valor_anterior, valor_nuevo = csv_handler.modificar(columna, "Iris-setosa-MODIFICADA", fila=0)

# Modificación 2: agregar un registro
nueva_fila = [151, 6.0, 3.0, 4.8, 1.8, "Iris-virginica"]
total_tras_agregar = csv_handler.agregar_registro(nueva_fila)

print("\n=== DESPUÉS (CSV) ===")
print(f"Fila 0 modificada: {csv_handler.datos[0]}")
print(f"Último registro agregado: {csv_handler.datos[-1]}")
print(f"Total de registros: {len(csv_handler.datos)}")

# Guardar nuevo CSV
csv_handler.guardar()
print("\n✅ Nuevo CSV guardado en: data/modificado.csv")

# Lo que se enviará al resumen (datos reales de la ejecución)
metodos_csv = ["importar", "modificar", "agregar_registro", "guardar"]
modificaciones_csv = [
    f"Actualizó '{columna}' en la fila 0: '{valor_anterior}' -> '{valor_nuevo}'",
    f"Agregó 1 registro: {nueva_fila} (total tras agregar: {total_tras_agregar})",
]
```

```python
# Inicializar e importar JSON
json_handler = ManejadorJSON('data/original.json', 'data/modificado.json')
json_handler.importar()
registros_antes_json = len(json_handler.datos)

print("=== ANTES (JSON) ===")
print(f"Primer registro: {json_handler.datos[0]}")
print(f"Total de registros: {registros_antes_json}")

# Modificación 1: agregar una clave a todos los registros
n_modificados = json_handler.modificar("procesado_por", "AE2_Evidencia_Aprendizaje")

# Modificación 2: eliminar un registro
registro_eliminado = json_handler.eliminar_registro(len(json_handler.datos) - 1)

print("\n=== DESPUÉS (JSON) ===")
print(f"Primer registro: {json_handler.datos[0]}")
print(f"Registro eliminado: {registro_eliminado}")
print(f"Total de registros: {len(json_handler.datos)}")

# Guardar nuevo JSON
json_handler.guardar()
print("\n✅ Nuevo JSON guardado en: data/modificado.json")

metodos_json = ["importar", "modificar", "eliminar_registro", "guardar"]
modificaciones_json = [
    f"Agregó la clave 'procesado_por' a {n_modificados} registros",
    f"Eliminó el último registro: {registro_eliminado}",
]
```

```python
# Inicializar la tercera clase con las mismas URLs usadas arriba
resumen_handler = GeneradorResumenTXT(URL_ORIGEN_CSV, URL_ORIGEN_JSON)

resumen_handler.registrar_operacion(
    formato="CSV",
    clase="ManejadorCSV",
    archivo_origen="data/original.csv",
    archivo_salida="data/modificado.csv",
    metodos=metodos_csv,
    modificaciones=modificaciones_csv,
    registros_antes=registros_antes_csv,
    registros_despues=len(csv_handler.datos),
)

resumen_handler.registrar_operacion(
    formato="JSON",
    clase="ManejadorJSON",
    archivo_origen="data/original.json",
    archivo_salida="data/modificado.json",
    metodos=metodos_json,
    modificaciones=modificaciones_json,
    registros_antes=registros_antes_json,
    registros_despues=len(json_handler.datos),
)

ruta_resumen = 'data/resumen.txt'
resumen_handler.generar_resumen(ruta_resumen)

print(f"✅ Resumen generado en: {ruta_resumen}\n")
print("=" * 40)
print("CONTENIDO DE resumen.txt:")
print("=" * 40)
with open(ruta_resumen, 'r', encoding='utf-8') as f:
    print(f.read())
```
